## Check 3 · Can you see a plot?
The next cell creates **fake (synthetic) data** and draws it. Nothing is downloaded: the data is generated by the computer.

**What you should see:**
* A scatter plot with **two clouds of points**: a **blue** one in the lower left and an **orange** one in the upper right.
* A **histogram** with two bumps (one per colour).
* The title of the figure says **"If you can read this, plots work"**.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(seed=42)          # fixed seed: everybody gets the same data
n = 200
data = pd.DataFrame({
    "group": ["A"] * n + ["B"] * n,
    "x": np.concatenate([rng.normal(2, 0.7, n), rng.normal(5, 0.7, n)]),
    "y": np.concatenate([rng.normal(2, 0.7, n), rng.normal(5, 0.7, n)]),
})
colors = {"A": "tab:blue", "B": "tab:orange"}

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))
for g, d in data.groupby("group"):
    ax1.scatter(d["x"], d["y"], s=12, alpha=0.7, color=colors[g], label=f"Group {g}")
    ax2.hist(d["x"], bins=20, alpha=0.7, color=colors[g])
ax1.set(xlabel="x", ylabel="y", title="Two clouds of points"); ax1.legend()
ax2.set(xlabel="x", ylabel="count", title="Histogram of x")
fig.suptitle("If you can read this, plots work")
plt.tight_layout()
plt.show()
```

## Check 4 · Can you see a table?
The next cell shows the first rows of the synthetic data and a summary. Group A should have an average `x` near **2** and group B near **5**.

```python
summary = data.groupby("group")[["x", "y"]].mean().round(2)
display(data.head())
display(summary)

ok = abs(summary.loc["A", "x"] - 2) < 0.3 and abs(summary.loc["B", "x"] - 5) < 0.3
print("✅ Numbers look right." if ok else "❌ Something is off: tell the instructor.")
```
