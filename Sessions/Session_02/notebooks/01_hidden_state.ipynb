# The hidden-state trap

A notebook remembers every variable you created, **even if you delete the cell that created it**. That is useful, and dangerous.

**Follow the instructions in order.**

### Step 1 · Run the next three cells in order (top to bottom)

```python
price = 10
```

```python
total = price * 3
print('total =', total)
```

```python
price = 20
print('price =', price)
```

### Step 2 · Now run **only** the `total = price * 3` cell again
What is `total` now? Is it consistent with the code you *see* on screen?

### Step 3 · Now go back to the cell `price = 20` and **change it to `price = 5`, but do NOT run it**.
Look at the notebook: which numbers are on the screen, and which would you get if you ran everything from scratch?

### Step 4 · The cure
In the toolbar click **Restart** and then **Run All** (or *Restart the kernel and run all cells*).
Rule: **before you share a notebook, restart and run all**. If it does not run from top to bottom, it is broken.

### Step 5 · A subtler version
Delete the next cell **after running it** and then run the one below: it still works. Is that good?

```python
secret_constant = 42
```

```python
print(secret_constant * 2)
```
