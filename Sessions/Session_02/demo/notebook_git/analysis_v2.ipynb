# Weekly sample analysis (v2: only the title changed!)

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng()
x = rng.normal(0, 1, 500)
```

```python
print('mean:', round(float(x.mean()), 3))
```

```python
plt.hist(x, bins=20)
plt.title('Histogram of x')
plt.show()
```
